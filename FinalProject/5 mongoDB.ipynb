```python
from pymongo import MongoClient
import pandas as pd
```

#### Connect to MongoDB

```python
# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")
```

```python
# Select the database iris_db
db = client['iris_db']
```

```python
# Select the database retail_db
collection = db['iris_data']
```

```python
# Retrieve the first three records
result = collection.find().limit(3)
# Print the records
for record in result:
    print(record)
```

```python
# Total number of species in this dataset
species_list = collection.distinct("species")
print("Unique species:", species_list)
print("Total number of species:", len(species_list))
```

```python
#show all rows has "setosa"
setosa_records = collection.find({"species": "Iris-setosa"})

print("All records with species 'setosa':")
for doc in setosa_records:
    print(doc)
```
